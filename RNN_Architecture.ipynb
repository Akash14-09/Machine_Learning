```python
from keras import Sequential
from keras.layers import Dense,SimpleRNN
```

```python
model = Sequential()
model.add(SimpleRNN(3,input_shape = (4,5)))
model.add(Dense(1,activation = 'sigmoid'))

model.summary()
```

model.get_weights()[0]  # First layer weights (5,3)


model.get_weights()[1]  # First layer biases (3,)


model.get_weights()[2]  # Second layer weights (3,3)


model.get_weights()[3]  # Second layer biases (3,)

```python
print(model.get_weights()[0].shape)
model.get_weights()[0]
#parameters of the array
#first leayer weights
```

```python
print(model.get_weights()[1].shape)
model.get_weights()[1]
#second layer weights
```

```python
print(model.get_weights()[2].shape)
model.get_weights()[2]
#second layer biases
```

```python
print(model.get_weights()[3].shape)
model.get_weights()[3]
#third layer weights
```

```python
print(model.get_weights()[4].shape)
model.get_weights()[4]
#third layer bias
```
