```python
import pandas as pd
import numpy as np
```

```python
from sklearn.linear_model import LinearRegression
```

```python
df = np.round(pd.read_csv('/content/50_Startups.csv')[['R&D Spend','Administration'  , 'Marketing Spend' , 'Profit']]/1000)
```

```python
df.head()
```

```python
np.random.seed(9)
```

```python
df = df.sample(5)
```

```python
df
```

```python
df = df.iloc[: , 0 : -1]
df
```

```python
df.iloc[1,0] = np.nan
df.iloc[3,1] = np.nan
df.iloc[-1,-1] = np.nan
```

```python
df.head()
```

```python
df0 = pd.DataFrame()

df0['R&D Spend'] = df['R&D Spend'].fillna(df['R&D Spend'].mean())
df0['Administration'] = df['Administration'].fillna(df['Administration'].mean())
df0['Marketing Spend'] = df['Marketing Spend'].fillna(df['Marketing Spend'].mean())
```

```python
df0
#0th iteration
```

```python
df1 = df0.copy()
df1.iloc[1,0] = np.nan
df1
```

Stage 1 --> When there is no missing values in the dataset
Note --> First prediction would be done witht the mean values then as we move further we keep predicting with the ml algorithm again and again and we keep changing our base dataset , by doing so we'll keep going more closer to the correct value


```python
#use first three rows to build a model and use the last row for the prediction

X = df1.iloc[[0,2,3,4] , 1: 3]
X
```

```python
y = df1.iloc[[0,2,3,4],0]
y
```

```python
lr = LinearRegression()

lr.fit(X,y)

lr.predict(df1.iloc[1,1:].values.reshape(1,2))
```

```python
df1.iloc[1,0] = 230.50
```

```python
df1
```
