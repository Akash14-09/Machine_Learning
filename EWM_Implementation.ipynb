```python
import pandas as pd


url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/daily-min-temperatures.csv"

df = pd.read_csv(url)
df.head()
```

```python
df = df[['Date' , 'Temp']]
df
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.scatter(df['Date'] , df['Temp'] , color = 'red')
```

```python
x1 = df['Temp'].ewm(alpha = 0.9 ).mean()
x1
```

```python
df['ewma'] = x1
df
```

```python
plt.scatter(df['Date'] , df['Temp'] , color = 'red')
plt.plot(df['Date'] , x1 , color = 'black' , label = "alpha = 0.9 or beta = 0.1")
plt.legend() ;
plt.show()
```
