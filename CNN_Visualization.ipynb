```python
from keras.applications.vgg16 import VGG16
model1 = VGG16()
```

```python
model1.summary()
```

```python
from keras.utils import plot_model
plot_model(model1)
```

```python
for i in range(len(model1.layers)):
  if 'conv' not in model1.layers[i].name:
    continue

  filters,biases = model1.layers[i].get_weights()
  print("layer number" , i , model1.layers[i].name , filters.shape)
```

```python
filters , bias = model1.layers[1].get_weights()
```

```python
f_min, f_max = filters.min(), filters.max()
filters = (filters - f_min) / (f_max - f_min)
```

```python
from matplotlib import pyplot

n_filters = 6

ix = 1

fig = pyplot.figure(figsize=(10,10))

for i in range(n_filters):
  f = filters[: , :  , : , i]
  for j in range(3):
    pyplot.subplot(n_filters , 3 , ix)
    pyplot.imshow(f[:,: , j] , cmap = 'gray')
    ix += 1
pyplot.show()
```

```python
from keras.models import Model
model = Model(inputs = model1.inputs , outputs = model1.layers[1].output)
```

```python
from keras.utils import load_img, img_to_array
from keras.applications.vgg16 import preprocess_input

image = load_img("/content/gettyimages-2221490610-612x612.jpg" , target_size = (224,224))
```

```python
image
```

```python
image = img_to_array(image)
image = image.reshape((1 , image.shape[0] , image.shape[1] , image.shape[2]))
image = preprocess_input(image)
```

```python
# calculating and plotting the feature maps in an 8x8 grid
features = model.predict(image)

# block1_conv1 has 64 filters, let's plot them as an 8x8 grid
square = 8
fig = pyplot.figure(figsize=(15, 15))

ix = 1
for i in range(square):
    for j in range(square):
        # Specify subplot and turn off axis ticks for clean visualization
        ax = pyplot.subplot(square, square, ix)
        ax.set_xticks([])
        ax.set_yticks([])

        # Plot the feature map (index ix-1)
        pyplot.imshow(features[0, :, :, ix - 1], cmap='gray')
        ix += 1

pyplot.tight_layout()
pyplot.show()
```

```python
model2 = VGG16()
```

```python
layer_index = [2,5,9,13,17]

outputs = [model2.layers[i].output for i in layer_index]
model3 = Model(inputs = model2.inputs , outputs = outputs)
```

```python
feature_map = model3.predict(image)
for i,fmap in zip(layer_index , feature_map) :
  fig = pyplot.figure(figsize=(20, 15))
  fig.suptitle("Layer_{}".format(i),fontsize = 20)
  for i in range(1,features.shape[3] + 1):
    pyplot.subplot(8,8,i)
    pyplot.imshow(fmap[0,:,:,i-1] , cmap = 'gray')
  pyplot.show()
```
