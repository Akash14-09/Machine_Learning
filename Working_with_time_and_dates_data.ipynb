##Working with Dates Data


```python
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

1.Extracting Year


```python
date['date_year'] = date['date'].dt.year
date.sample(5)
```

2.Extracting the month from the dataset

```python
date['date_month_no'] = date['date'].dt.month_name()
date.head()
```

3.Extracting days


```python
date['date_day'] = date['date'].dt.day
date.head()
```

```python
date['date_dow'] = date['date'].dt.dayofweek
date.head()
```

```python
date['date_dow_name'] = date['date'].dt.day_name()

date.drop(columns = ['product_id' , 'city_id' , 'orders']).head()
```

```python
date.head()
```

```python
date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday' , 'Saturday']) , 1 , 0)

date.drop(columns = ['product_id' , 'city_id' , 'orders']).head()
```

Extracting week of the year


```python
date['date_week'] = date['date'].dt.isocalendar().week

date.drop(columns = ['product_id' , 'city_id' , 'orders']).head()
```

Extract the quarter


```python
date['quarter'] = date['date'].dt.quarter
date.drop(columns = ['product_id' , 'city_id' ,'orders']).head()
```

Extract Semester


```python
date['semester'] = np.where(date['quarter'].isin([1,2] ) , 1 , 2)
date.drop(columns = ['product_id' , 'city_id' , 'orders']).head()
```

Extract the time elapsed betweeen the dates

```python
import datetime
today = datetime.datetime.today()
today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```

month passed

```python
np.round((today - date['date']).dt.days / 30.44 , 0)
```
