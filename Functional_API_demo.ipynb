```python
x = Input(shape  = (3, ))
```

```python

hidden1 = Dense(128 , activation= 'relu')(x)
#hidden1 layer is getting input from x
hidden2 = Dense(64 , activation='relu')(hidden1)

output1  = Dense(1, activation='linear')(hidden2)
output2 = Dense(1,activation='sigmoid')(hidden2)
```

##Multiple Inputs to One Output

```python
inputA = Input(shape=(32,))
inputB = Input(shape=(128,))
```

```python
x1 = Dense(8, activation="relu")(inputA)
x2 = Dense(4, activation="relu")(x1)
#the first branch operates in the first input
```

```python
#the second branch operates on the second input
y1 = Dense(64, activation="relu")(inputB)
y2 = Dense(32, activation="relu")(y1)
y3 = Dense(8, activation="relu")(y2)
```

```python
#now combining the ouptut of the both branches
combined = concatenate([x2, y3])
```

```python
#now applying a fully connnected layer and hten a regression prediction on the combined outputs

z = Dense(2, activation="relu")(combined)
z1 = Dense(1, activation="linear")(z)
```

```python
#our model will acccept the output from both the branches and then output a single value
model2 = Model(inputs=[inputA, inputB], outputs=z1)
```

```python
from keras.utils import plot_model

plot_model(
    model2,
    show_shapes=True,
    show_layer_names=True,
    rankdir="LR",
    dpi=70
)
```
